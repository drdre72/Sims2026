# NetworkX getting started

Sims2026 starter notebook. `import networkx` builds graphs, runs algorithms,
and pairs with matplotlib for drawing.

```python
import networkx as nx
import matplotlib.pyplot as plt
print(nx.__version__)
```

```python
# Build a small graph
G = nx.Graph()
G.add_edges_from([("ALU", "REG"), ("REG", "PC"), ("PC", "MEM"),
                  ("MEM", "ALU"), ("ALU", "CTRL"), ("CTRL", "PC")])
print(G.number_of_nodes(), "nodes,", G.number_of_edges(), "edges")
print("degree of ALU:", G.degree["ALU"])
print("shortest PC -> CTRL:", nx.shortest_path(G, "PC", "CTRL"))
```

```python
# Draw it
pos = nx.spring_layout(G, seed=7)
nx.draw(G, pos, with_labels=True, node_color="lightblue",
        node_size=1500, font_weight="bold")
plt.show()
```

```python
# Classic algorithms
print("clustering:", nx.clustering(G))
print("pagerank top-3:", sorted(nx.pagerank(G).items(),
      key=lambda kv: -kv[1])[:3])
```
